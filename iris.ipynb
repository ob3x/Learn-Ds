```python
import numpy as np
import pandas as pd
import sklearn
import seaborn as sns
```

```python
np.random.seed(42)
np.set_printoptions(edgeitems=30, linewidth=100000, formatter=dict(float=lambda x: f"{x:.2f}"))

sns.set()
```

```python
from sklearn.datasets import load_iris

raw_df = load_iris()
raw_df_copy = raw_df.copy()

x = raw_df_copy["data"]
y = raw_df_copy["target"]

raw_df_copy.keys()
```

```python
data = pd.DataFrame(data=x, columns=raw_df_copy["feature_names"])
target = pd.DataFrame(data=y, columns=["target"])
df = pd.concat([data, target], axis=1)
df
```

```python
df.describe().T.apply(lambda x: round(x, 2))
```

```python
df.info()
```

```python
df.target.value_counts().plot(kind="pie")
df.target.value_counts()
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(data, target, test_size=0.25, random_state=42)

print(f"x_train shape : {x_train.shape}")
print(f"y_train shape : {y_train.shape}")
print(f"x_test shape : {x_test.shape}")
print(f"y_test shape : {y_test.shape}")

print(f"\ny_train value : {y_train.value_counts()}")
print(f"\ny_test value : {y_test.value_counts()}")
```

```python
from sklearn.datasets import load_breast_cancer

raw_data = load_breast_cancer()
raw_data_copy = raw_data.copy()
raw_data_copy.keys()
```

```python
x = raw_data_copy["data"]
y = raw_data_copy["target"]

data = pd.DataFrame(x, columns=raw_data_copy["feature_names"])
target = pd.DataFrame(y, columns=["target"])
df = pd.concat([data, target], axis=1)
df.info()
```

```python
df.describe().T.apply(lambda x: round(x, 2))
```

```python
df.target.value_counts()
```

```python
df.target.value_counts().plot(kind="pie")
```
