```python
import kagglehub
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import sklearn

np.random.seed(42)


# Download latest version
path = kagglehub.dataset_download("sergionefedov/speed-dating-and-attraction-outcomes")

print("Path to dataset files:", path)

raw_data = pd.read_csv(f"{path}/speed_dating_master.csv")
raw_data
```

```python
df = raw_data.copy()
df.info()
```

```python
df.describe().T
```

```python
df.isnull().sum().sum()
```

```python
for col in ["event_id", "male_id", "intel_of_female", "fun_of_female", "amb_of_female", "female_id", "match", "male_decision", "female_goes_out", "female_self_attr", "female_pref_intel", "female_pref_attr", "sinc_of_female", "attr_of_female", "female_age", "age_gap"]:
  df.drop(col, axis=1, inplace=True)

df.head()
```

```python
target = df["female_decision"]
data = df.drop("female_decision", axis=1)
data.head()
```

```python
data = pd.get_dummies(data, drop_first=True, dtype=int)
data.head()
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(data, target, test_size=0.2, random_state=42)

print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(9, 5))

# Wykres gęstości (KDE) - jak rozkłada się atrakcyjność dla "Tak" vs "Nie"
sns.kdeplot(
    data=df,
    x='attr_of_male',
    hue='female_decision',
    common_norm=False,
    fill=True,
    palette={0: '#d95f02', 1: '#2b5c8f'}
)

plt.title('Rozkład ocen atraktywności a decyzja kobiety')
plt.xlabel('Ocena atraktywności mężczyzny (oryginalna skala)')
plt.ylabel('Gęstość')
plt.show()
```

```python
from sklearn.linear_model import LogisticRegression

log_reg = LogisticRegression(class_weight="balanced")
log_reg.fit(X_train, y_train)
```

```python
log_reg.score(X_test, y_test)
```

```python
y_pred = log_reg.predict(X_test)
y_comparison = pd.DataFrame({"y_true" : y_test, "y_pred" : y_pred})
y_comparison
```

```python
from sklearn.metrics import classification_report, confusion_matrix

y_pred = log_reg.predict(X_test)

print("--- Macierz pomyłek (Confusion Matrix) ---")
print(confusion_matrix(y_test, y_pred))

print("\n--- Raport Klasyfikacji ---")
print(classification_report(y_test, y_pred))
```
