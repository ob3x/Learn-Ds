```python
import kagglehub
import numpy as np
import pandas as pd
import sklearn
import matplotlib.pyplot as plt

# Download latest version
path = kagglehub.dataset_download("andonians/random-linear-regression")

print("Path to dataset files:", path)

train_data = pd.read_csv(f"{path}/train.csv")
test_data = pd.read_csv(f"{path}/test.csv")
```

```python
train_data = train_data.dropna()
```

```python
x_train = train_data[["x"]]
y_train = train_data["y"]
x_test = test_data[["x"]]
y_test = test_data["y"]
```

```python
plt.figure(figsize=(8, 6))
plt.title("dane treningowe vs dane testowe")
plt.xlabel("cecha x")
plt.ylabel("cecha y")
plt.scatter(x_train, y_train, c="r", alpha=0.3, label="dane treningowe")
plt.scatter(x_test, y_test, c="b", alpha=0.3, label="dane testowe")
plt.legend()
plt.show()
```

```python
from sklearn.linear_model import LinearRegression

lr = LinearRegression()
lr.fit(x_train, y_train)
```

```python
lr.score(x_test, y_test)
```

```python
y_pred = lr.predict(x_test)
y_pred[:5]
```

```python
from sklearn.metrics import r2_score

r2_score(y_test, y_pred)
```

```python
y_comparison = pd.DataFrame({"y_test": y_test, "y_pred": y_pred})
y_comparison.head()
```

```python
error = y_comparison["y_test"] - y_comparison["y_pred"]
y_comparison["error"] = error
y_comparison.head()
```

```python
plt.figure(figsize=(8, 6))
plt.title("wykres prostej")
plt.xlabel("cecha x")
plt.ylabel("cecha y")
plt.scatter(x_test, y_test, c="b", alpha=0.3, label="dane testowe")
plt.plot(x_test, lr.intercept_ + x_test * lr.coef_[0], c="r")
plt.legend()
plt.show()
```
