```python
url_1 = "https://images.unsplash.com/photo-1506744038136-46273834b3fb"
url_2 = "https://images.unsplash.com/photo-1794246561495-d69e5d7d5668?q=80&w=963&auto=format&fit=crop&ixlib=rb-4.1.0&ixid=NHwxMjA3fDA%3D"


img_1 = iio.imread(url_1)
img_2 = iio.imread(url_2)

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 2)
plt.imshow(img_1)
plt.axis("off")
plt.title("Zdjęcie 1")

plt.subplot(1, 2, 1)
plt.imshow(img_2)
plt.axis("off")
plt.title("Zdjęcie 2")
```

```python
def mostly_red(x):

    is_red = (x[:, :, 0] > x[:, :, 1]) & (x[:, :, 0] > x[:, :, 2])

    print(is_red.sum(), x.size / 6)

    if is_red.sum() >= x.size / 6:
        return True

    return False

print(mostly_red(img_1))
print(mostly_red(img_2))
```
